# **📋 Encuestas: Conceptos Fundamentales**

> ## **¿Qué es una encuesta?**

Una **encuesta** es una *técnica de recolección de información* que consiste en formular un conjunto de preguntas a una *muestra* o a toda una *población* con el propósito de obtener datos sobre *opiniones*, *comportamientos*, *características*, *conocimientos* o *preferencias*.

Las encuestas constituyen una de las herramientas más utilizadas en la *estadística*, la *investigación científica*, el *mercadeo*, la *economía*, las *ciencias sociales*, la *salud pública* y la *administración*, ya que permiten obtener información de manera rápida, sistemática y organizada.

---

# **🎯 Importancia de las encuestas**

Las encuestas permiten:

* ✅ *Conocer* las opiniones y necesidades de una población.
* ✅ *Apoyar* la toma de decisiones basada en datos.
* ✅ *Evaluar* la satisfacción de clientes, estudiantes o ciudadanos.
* ✅ *Estimar* parámetros poblacionales mediante muestras.
* ✅ *Identificar* tendencias y cambios en el comportamiento de las personas.
* ✅ *Formular* políticas públicas y estrategias empresariales.

---

## **📌 Ejemplo**

Una universidad desea conocer el *nivel de satisfacción* de sus estudiantes con los servicios de biblioteca.

En lugar de entrevistar a los **20 000 estudiantes**, selecciona una **muestra** de **600 estudiantes** y aplica una **encuesta**.

A partir de esta muestra es posible *estimar* la satisfacción de toda la *población estudiantil*.

---

# **📊 Clasificación de las encuestas**

Las encuestas pueden clasificarse de diferentes maneras según el *medio de aplicación*, el *tiempo*, el *objetivo*, el *tipo de preguntas* y el *alcance del estudio*.

---

# **1. Según el medio de aplicación**

| **Tipo**                | **Descripción**                                                       | **Ejemplo**                    |
| ----------------------- | --------------------------------------------------------------------- | ------------------------------ |

| **Longitudinal** | Se aplica repetidamente para estudiar los cambios a través del tiempo.        |

### **Ejemplo**

* **Transversal:** Encuesta sobre desempleo en julio de 2026.
* **Longitudinal:** Encuesta aplicada cada año durante diez años para analizar la evolución del desempleo.

# **4. Según el tipo de preguntas**

## **🔹 Preguntas abiertas**

Permiten que el participante responda con sus propias palabras.

**Ejemplo**

> ¿Qué aspectos considera que deberían mejorar en la universidad?

---

## **🔹 Preguntas cerradas**

El participante selecciona una respuesta previamente establecida.

**Ejemplo**

**¿Cuál es su nivel de satisfacción?**

* Muy satisfecho
* Satisfecho
* Neutral
* Insatisfecho
* Muy insatisfecho

---

## **🔹 Preguntas mixtas**

Combinan *preguntas cerradas* con un espacio para que el encuestado pueda realizar comentarios adicionales.

# **5. Según el alcance**

| **Tipo**     | **Descripción**                                                        |
| ------------ | ---------------------------------------------------------------------- |
| **Censal**   | Se encuesta a **todos** los elementos de la población.                 |
| **Muestral** | Se encuesta únicamente una **muestra representativa** de la población. |

# **📝 Etapas para realizar una encuesta**

1. **Definir** el *objetivo* del estudio.
2. **Identificar** la *población objetivo*.
3. **Diseñar** el *cuestionario*.
4. **Seleccionar** el *método de muestreo*.
5. **Aplicar** la encuesta.
6. **Procesar** y depurar los datos.
7. **Analizar** los resultados.
8. **Elaborar** conclusiones y recomendaciones.

---

# **⚠️ Factores que afectan la calidad de una encuesta**

La calidad de una encuesta depende de:

* *Diseño adecuado* del cuestionario.
* *Selección correcta* de la muestra.
* *Cobertura* de la población objetivo.
* *Honestidad* de las respuestas.
* *Alta tasa de respuesta*.
* *Capacitación* de los encuestadores.
* *Procesamiento y análisis* adecuado de los datos.

Cuando alguno de estos aspectos falla, pueden aparecer **errores** o **sesgos** que afectan la *validez* y la *confiabilidad* de los resultados.

---

# **🔄 Transición hacia el estudio de los sesgos**

Una encuesta bien diseñada debe producir información **válida**, **confiable** y **representativa** de la población. Sin embargo, durante su planificación, aplicación o análisis pueden introducirse **sesgos**, es decir, *errores sistemáticos* que distorsionan los resultados y pueden conducir a conclusiones equivocadas.

En la siguiente sección estudiaremos los **principales tipos de sesgos**, cómo identificarlos y las estrategias para prevenirlos.

---

> **💡 Idea clave:** Una **encuesta** es una herramienta fundamental para la toma de decisiones basada en datos. No obstante, su utilidad depende de que el proceso de diseño, selección de la muestra, formulación de las preguntas y recolección de la información se realice con rigor metodológico, minimizando la presencia de **errores** y **sesgos**.



# **⚠️ Calidad de las encuestas: errores y sesgos**

Una **encuesta** bien diseñada debe proporcionar información *válida*, *confiable* y *representativa* de la **población objetivo**. Sin embargo, durante el diseño del cuestionario, la selección de la muestra, la recolección de los datos o el análisis de la información pueden introducirse **errores sistemáticos**, conocidos como **sesgos**, que afectan la calidad de los resultados.

Es importante destacar que un **sesgo** no desaparece aumentando el tamaño de la muestra. Si el diseño de la encuesta es deficiente, los resultados seguirán siendo incorrectos, aunque se encueste a un gran número de personas.

---

# **📌 ¿Qué es un sesgo?**

Un **sesgo** es un *error sistemático* que provoca que los resultados de una encuesta se alejen del verdadero valor de la población. Los sesgos pueden conducir a conclusiones equivocadas y afectar la toma de decisiones.

**Ejemplo**

Si una encuesta pretende medir el nivel de satisfacción de los estudiantes de una universidad, pero únicamente participan aquellos que asisten regularmente a clases, los resultados podrían **sobreestimar** el nivel real de satisfacción.

---

# **🎯 Principales tipos de sesgo en las encuestas**

## **1. Sesgo de respuesta**

### **Definición**

Ocurre cuando los *encuestados no responden con total honestidad* o modifican sus respuestas por motivos sociales, culturales o personales.

### **Ejemplo**

Una encuesta pregunta:

> **¿Cuántos días hace ejercicio por semana?**

Algunas personas responden **cinco días**, aunque realmente solo hacen ejercicio **dos veces**, porque desean proyectar una imagen más saludable.

### **Consecuencia**

Se *sobreestima* el nivel de actividad física de la población.

---

## **2. Sesgo de no respuesta**

### **Definición**

Se presenta cuando *ciertos grupos de la población no responden* la encuesta o lo hacen en menor proporción que otros.

### **Ejemplo**

Una empresa realiza una encuesta sobre satisfacción laboral.

Los empleados más inconformes deciden no responder por temor a represalias.

### **Consecuencia**

Los resultados muestran un nivel de satisfacción mayor al real.

---

## **3. Sesgo por formulación de preguntas**

### **Definición**

Ocurre cuando las *preguntas están redactadas de forma que influyen en la respuesta* del encuestado.

### **Pregunta sesgada**

> **¿No cree que una dieta vegana es más ética que consumir carne?**

### **Pregunta neutral**

> **¿Cuál es su opinión sobre la dieta vegana?**

### **Consecuencia**

Las respuestas reflejan la influencia de la pregunta y no necesariamente la opinión real del participante.

---

# **📱 Cobertura insuficiente: una fuente importante de sesgo**

La **cobertura insuficiente** ocurre cuando *una parte de la población objetivo no tiene oportunidad de ser seleccionada* para participar en la encuesta.

## **Ejemplo**

Una empresa realiza una encuesta telefónica utilizando **únicamente teléfonos fijos**.

### **¿Qué ocurre?**

Se excluyen personas que solo utilizan teléfonos móviles, especialmente jóvenes y algunos trabajadores.

### **Consecuencia**

La muestra deja de ser representativa y los resultados reflejan principalmente la opinión de personas de mayor edad.

> **Conclusión:** Una cobertura incompleta produce una **muestra sesgada**, lo que afecta la validez de las conclusiones.

---

# **🔍 ¿Cómo identificar un sesgo en una encuesta?**

## **1. Comparar la muestra con la población objetivo**

Se debe verificar si la muestra mantiene una distribución similar a la población en variables como *sexo*, *edad*, *nivel educativo*, *ocupación* o *región*.

### **Ejemplo**

| **Población** | **Muestra**  |
| ------------- | ------------ |
| 50 % Hombres  | 70 % Hombres |
| 50 % Mujeres  | 30 % Mujeres |

La muestra **no representa adecuadamente** a la población, por lo que existe un **sesgo de selección**.

---

## **2. Revisar la redacción de las preguntas**

Las preguntas deben ser *claras*, *neutrales* y *objetivas*.

### **Pregunta sesgada**

> **¿Qué tan de acuerdo está con que los impuestos son demasiado altos?**

### **Pregunta neutral**

> **¿Cuál es su opinión sobre el nivel actual de los impuestos?**

---

## **3. Analizar los patrones de no respuesta**

Es importante verificar si determinados grupos participan menos que otros.

### **Ejemplo**

* Jóvenes (18–24 años): **30 %** de respuesta.
* Adultos mayores: **90 %** de respuesta.

En este caso, los resultados reflejarán principalmente la opinión de los adultos mayores.

---

# **✅ Estrategias para recopilar datos sin sesgo**

## **1. Utilizar métodos de muestreo probabilístico**

Seleccionar la muestra mediante técnicas como el *muestreo aleatorio simple*, *estratificado*, *sistemático* o *por conglomerados*, garantizando que todos los individuos tengan una probabilidad conocida de ser seleccionados.

---

## **2. Diseñar preguntas neutrales**

Las preguntas deben:

* Ser *claras* y *precisas*.
* Evitar palabras que induzcan una respuesta.
* No contener juicios de valor.
* Formular una sola idea por pregunta.

---

## **3. Garantizar una cobertura adecuada**

Todos los subgrupos de la población deben tener la oportunidad de participar.

Por ejemplo, combinar encuestas:

* Presenciales.
* Telefónicas.
* En línea.
* Mediante dispositivos móviles.

---

## **4. Realizar una prueba piloto**

Antes de aplicar la encuesta definitiva, se recomienda realizar una *prueba piloto* con un grupo reducido de participantes para detectar problemas de comprensión, redacción o logística.

---

## **5. Corregir el sesgo mediante ponderaciones**

Cuando algunos grupos están *subrepresentados*, es posible aplicar **ponderaciones estadísticas** para aproximar la muestra a la distribución real de la población.

### **Ejemplo**

Si los jóvenes representan el **40 %** de la población, pero solo constituyen el **20 %** de la muestra, sus respuestas pueden recibir un mayor peso durante el análisis estadístico.

---

# **💡 Ejemplo integrador**

Una universidad desea conocer la satisfacción de sus estudiantes.

**Diseño inicial**

* La encuesta se envía únicamente al correo institucional.
* Responden principalmente estudiantes de Ingeniería.

**Problemas identificados**

* ❌ *Cobertura insuficiente*.
* ❌ *Sesgo de no respuesta*.
* ❌ *Muestra desbalanceada* entre programas académicos.

**Acciones correctivas**

* Enviar la encuesta por correo, plataforma virtual y WhatsApp.
* Aplicar un *muestreo estratificado* por programa académico.
* Diseñar preguntas *claras* y *neutrales*.
* Realizar una *prueba piloto*.
* Aplicar *ponderaciones* cuando sea necesario.

**Resultado**

Se obtiene una muestra más representativa y, por tanto, resultados **más válidos**, **más confiables** y útiles para la toma de decisiones.

---

# **📚 Ideas clave**

> * Una **encuesta** solo es útil si proporciona información *válida* y *representativa*.
> * Un **sesgo** es un *error sistemático* que puede aparecer durante el diseño, la aplicación o el análisis de una encuesta.
> * Los principales sesgos son el *sesgo de respuesta*, el *sesgo de no respuesta*, el *sesgo por formulación de preguntas* y el *sesgo por cobertura insuficiente*.
> * La mejor estrategia para minimizar los sesgos consiste en diseñar cuidadosamente la encuesta, utilizar un **muestreo probabilístico**, redactar preguntas *neutrales*, realizar una *prueba piloto* y verificar continuamente la representatividad de la muestra.
